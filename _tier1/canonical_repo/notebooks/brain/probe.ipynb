# Dataset attachment probe
Quickly determines whether `videlisndichi/silverwing-120m` is mounted into the kernel and at what path. No GPU, no torch install.

```python
import os
print('=== dataset attachment probe ===')
print('os.listdir(/kaggle/input):', os.listdir('/kaggle/input') if os.path.isdir('/kaggle/input') else 'NO /kaggle/input')
found = []
if os.path.isdir('/kaggle/input'):
    for root, dirs, files in os.walk('/kaggle/input'):
        for f in files:
            found.append(os.path.relpath(os.path.join(root, f), '/kaggle/input'))
print('total files under /kaggle/input:', len(found))
for x in sorted(found):
    print('  ', x)
have_train_py = any('silverwing_train.py' in f for f in found)
have_corpus = any('01_normalized.jsonl' in f for f in found)
print('silverwing_train.py present:', have_train_py)
print('01_normalized.jsonl present:', have_corpus)
print('PROBE RESULT:', 'DATASET ATTACHED' if have_train_py else 'DATASET NOT ATTACHED')
```
